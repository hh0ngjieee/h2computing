**Task 2.2**

Write a function `task2_2(list_of_integers)` that:

- takes a list of integers, `list_of_integers`
- sorts them into ascending order using merge sort
- returns the sorted list. **[7]**

Use the list `[56,25,4,98,0,18,4,5,7,0]` to test your function.

For example, the condition

`task2_2([56,25,4,98,0,18,4,5,7,0]) == [0,0,4,4,5,7,18,25,56,98]`

should return `True`. **[2]**


````python
# Task 2.2
def task2_2(list_of_integers):

    if len(list_of_integers) == 1:
        return list_of_integers

    #split the list into 2 halves
    list1 = list_of_integers[0:int(len(list_of_integers)/2)]
    list2 = list_of_integers[int(len(list_of_integers)/2):len(list_of_integers)]

    #recursively sort the lists
    list1 = task2_2(list1)
    list2 = task2_2(list2)

    #create a new list
    sorted_list = []
    i, j = 0, 0

    #add in the contents of the list
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            sorted_list.append(list1[i])
            i += 1
        else:
            sorted_list.append(list2[j])
            j += 1

    #this runs only if list 2 is entirely wiped out
    while i < len(list1):
        sorted_list.append(list1[i])
        i += 1

    #this runs only if list 1 is entirely wiped out
    while j < len(list2):
        sorted_list.append(list2[j])
        j += 1

    return sorted_list

print(task2_2([56,25,4,98,0,18,4,5,7,0]) == [0,0,4,4,5,7,18,25,56,98])
````

**Task 2.3**

Write a function `task2_3(filename_in, filename_out)` that:

- accepts two parameters:
  - `filename_in` represents the input file name
  - `filename_out` represents the output file name
- reads the integers from the input file
- uses your `task2_2` function to sort the integers
- writes the integers to the output file. **[5]**

The function should read the random numbers from

`randomnumbers_<your name>_<centre number>_<index number>.txt`

and then write the sorted integers to

`sortednumbers_<your name>_<centre number>_<index number>.txt` **[3]**


````python
# Task 2.3
from random import *

def task2_3(filename_in, filename_out):

    #create file which automatically closes
    with open(filename_in, "w") as file:

        #create an array of a reasonable size with integers of reasonable values, in that case array size 10 and integers from 0 to 100
        for i in range(10):
            file.write(str(randint(0,100)) + "\n")

    list_of_integers = []

    #read the file and obtain the integers, file automatically closes
    with open(filename_in, "r") as file:

        #add in integers for each line
        for line in file:
            line.strip()
            list_of_integers.append(int(line))

    #call task2_2 with the unsorted list
    new_list = task2_2(list_of_integers)

    #insert items into new file, file closes automatically
    with open(filename_out, "w") as file:

        #add the numbers in the sorted list one by one
        for item in new_list:
            file.write(str(item) + "\n")

task2_3("randomnumbers.txt", "sortednumbers.txt")
````
